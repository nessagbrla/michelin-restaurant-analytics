# Bronze Layer: Michelin Restaurant Data

This notebook documents the ingestion of the raw Michelin restaurant dataset into the Bronze layer.

The Bronze layer preserves source data in its original form without applying business transformations. Cleaning, standardization, and derived fields are handled in the Silver layer.

Source: Public Michelin restaurant dataset
Raw records: 19,622

```sql

SELECT COUNT(*) AS bronze_row_count
FROM workspace.michelin_analytics.bronze_michelin_restaurants;
```

```sql
DESCRIBE TABLE workspace.michelin_analytics.bronze_michelin_restaurants;
```

### Basic Ingestion Validation

```sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT CONCAT(
        Name, '|',
        Address, '|',
        Location, '|',
        Latitude, '|',
        Longitude
    )) AS unique_restaurant_locations
FROM workspace.michelin_analytics.bronze_michelin_restaurants;
```
