# Exploratory Analysis: Global Michelin Restaurants

This notebook explores the analytics-ready Michelin restaurant dataset to identify patterns in geographic concentration, Michelin recognition, cuisine, price positioning, and Green Star recognition.

The analysis is used to determine the key insights and visualizations presented in the final Power BI dashboard.

## Question 1: Which countries have the largest Michelin presence?

```sql
SELECT
    country,
    COUNT(*) AS restaurant_count,
    SUM(CASE WHEN star_count >= 1 THEN 1 ELSE 0 END) AS starred_restaurants,
    SUM(CASE WHEN star_count = 3 THEN 1 ELSE 0 END) AS three_star_restaurants,
    SUM(CASE WHEN is_green_star THEN 1 ELSE 0 END) AS green_star_restaurants,
    ROUND(
        100.0 * SUM(CASE WHEN star_count >= 1 THEN 1 ELSE 0 END) / COUNT(*),
        1
    ) AS starred_percentage
FROM workspace.michelin_analytics.gold_restaurant_analytics
GROUP BY country
ORDER BY restaurant_count DESC;
```

## Question 2: Which cities have the largest Michelin restaurant markets?

```sql
SELECT
    city,
    country,
    COUNT(*) AS restaurant_count,

    SUM(
        CASE WHEN star_count >= 1 THEN 1 ELSE 0 END
    ) AS starred_restaurants,

    ROUND(
        100.0 *
        SUM(CASE WHEN star_count >= 1 THEN 1 ELSE 0 END)
        / COUNT(*),
        1
    ) AS starred_percentage,

    SUM(
        CASE WHEN star_count = 3 THEN 1 ELSE 0 END
    ) AS three_star_restaurants,

    ROUND(AVG(price_tier), 2) AS avg_price_tier

FROM workspace.michelin_analytics.gold_restaurant_analytics

GROUP BY city, country

HAVING COUNT(*) >= 10

ORDER BY restaurant_count DESC

LIMIT 30;
```

## Question 3: Which cuisines have the strongest Michelin presence?

```sql
SELECT
    cuisine,
    COUNT(*) AS restaurant_count,
    SUM(CASE WHEN star_count >= 1 THEN 1 ELSE 0 END) AS starred_restaurants,
    SUM(CASE WHEN star_count = 3 THEN 1 ELSE 0 END) AS three_star_restaurants,
    ROUND(
        100.0 * SUM(CASE WHEN star_count >= 1 THEN 1 ELSE 0 END) / COUNT(*),
        1
    ) AS starred_percentage,
    ROUND(AVG(price_tier), 2) AS avg_price_tier
FROM workspace.michelin_analytics.gold_restaurant_analytics
GROUP BY cuisine
HAVING COUNT(*) >= 30
ORDER BY restaurant_count DESC;
```

## Key Findings

- Michelin restaurant coverage is concentrated in a relatively small number of major markets, led by France, Italy, the United States, Spain, Germany, Japan, and the United Kingdom.
- Restaurant volume and Michelin star concentration do not necessarily move together. Japan has fewer Michelin-listed restaurants than several European markets but a comparatively high share of starred restaurants.
- Tokyo has the largest restaurant count among cities in the dataset, followed by Paris and London.
- Cuisine categories show substantial differences in Michelin recognition. Large categories such as Modern Cuisine have significant representation, while categories such as Creative cuisine show particularly high concentrations of starred restaurants.
- Higher star concentration is observed in several cuisine categories with higher average price tiers, although this analysis establishes association rather than causation.
- Comparisons based on percentages should account for sample size, particularly for countries, cities, and cuisines with relatively few restaurants.
